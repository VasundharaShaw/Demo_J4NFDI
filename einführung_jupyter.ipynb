# Einführung in JupyterNotebooks

## Was JupyterNotebooks sind 🧮
Ein Jupyter Notebook ist ein interaktives Dokument, das Code (oft Python), Text (Markdown, so wie dieses Format hier), Mathematik (LaTeX) und Visualisierungen in einer einzigen Benutzeroberfläche kombiniert. Es wird vor allem in Datenwissenschaft, Forschung, Bildung und Maschinellem Lernen verwendet.

* Unterstützt viele Programmiersprachen (vor allem Python, aber auch R, Julia, etc.).

* Besteht aus sogenannten „Zellen“ – man kann Code- und Textabschnitte getrennt ausführen.

* Ergebnisse (z. B. Tabellen, Grafiken) erscheinen direkt unter dem Code.

📁 Dateiformat: .ipynb (JSON-basiert)

## Weiterführende Informationen
Außer JupyterNotebooks gibt es auch JupyterLabs und JupyterHubs. Es ist nicht besonders wichtig alles im Detail zu kennen, aber es kann trotzdem für Sie interessant sein, was die Begriffe bedeuten. Daher können Sie gerne weiterlesen ;) .

### JupyterLabs 🧪
JupyterLab ist die weiterentwickelte Benutzeroberfläche für Jupyter Notebooks – eine Art "IDE für Datenarbeit". Im Grunde ist der aktuelle Bereich ein JupyterLab.

* Bietet eine moderne, flexible Oberfläche mit Tabs, Drag & Drop und Panels.

* Unterstützt Notebooks, Terminals, Texteditoren, Dateibrowser, Plot-Viewer etc. in einer Umgebung.

* Macht die Arbeit mit mehreren Dokumenten gleichzeitig deutlich komfortabler.

🆚 Vergleich zu Jupyter Notebook:

* Jupyter Notebook ist einfach und direkt – JupyterLab ist vielseitiger und moderner.

![jupyter_lab_example](assets/jupyter_lab.png)

### JupyterHub 🧑‍🤝‍🧑
JupyterHub ist ein Mehrbenutzersystem, das Jupyter Notebooks oder JupyterLab zentral für viele Nutzer bereitstellt – z. B. in Schulen, Universitäten, Forschungseinrichtungen oder Firmen.

* Bietet Benutzerauthentifizierung (z. B. Login mit LDAP, GitHub, Google, etc.).
    * Sie haben sich eventuell mit dem Account Ihrer Universität angemeldet.

* Startet für jeden Benutzer eine eigene Notebook-Umgebung (oft in Containern oder per SSH).

* Die Nutzer haben direkter Zugriff auf die Funktionen
    * Man muss direkt nicht selber etwas installieren
    * Keine Administration für die Umgebung vom Nutzer notwendig

🛠 Beispiel-Szenarien:

* Ein Kurs für die Einführung in die Web-Entwicklung, in dem alle Studierenden dieselbe Umgebung nutzen.

* Forschungscluster, bei dem die Benutzer auf leistungsfähige Rechenknoten verteilt werden und man auf verschiedenste Datensätze zugreifen kann.
    * Ressourcen teilen
    * Daten teilen
    * Vielen Forschenden Kapazitäten eröffnen
