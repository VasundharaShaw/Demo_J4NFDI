## 2️⃣ 🔁 JupyterLab XL
Different images with different functionalities. The [standard JupyterLab](https://nfdi-jupyter.de/users/jupyterlab/default/) is based on the jupyter/minimal-notebook. This includes a python kernel. 
The [JupyterLab XL](https://nfdi-jupyter.de/users/jupyterlab/4.3/) gives access to a lot of additional functions. You can select it via "Kernels and Extensions" in the configuration of JupyterLab:
* Different Extensions: JupyterView, HPC Terminals, ...
* Kernels: Bash, Java, Julia, ...
* Proxies: Xpra, VSCode, ...
<img src="assets/kernel_and_extensions_example.png" alt="Kernel and Extensions" height=3200 width=800 >

### 👉 Look at all possibilities
Now we have a look at the JupyterLab XL with all selectable options. This may take a while to build.
* In the JupyterLab Overview add a new JupyterLab.
* Select JupyterLab XL at **Select Version** section.
* Change the tab on the right side to **Kernel and Extensions**.
* Then click on **Select all**.
* Start the JupyterLab by clicking on the **Start** Button.
<img src="assets/jupyterlab_xl_example.png" alt="JupyterLab XL example" height=3200 width=800 >

* Now you can close this JupyterLab again.

## 3️⃣  🔁 Repo2Docker – GitHub Integration
We will now explore the Repo2Docker feature. This feature allows you to integrate many repository into a JupyterLab environment. It also lets you specify, in separate files, which software packages are required for the project contained within the repository. In this section, we will briefly discuss this feature and demonstrate how it works.
In the **Select Version** area it is possible to use the Repo2Docker functionality. You can select a repository. This will be imported in a JupyterLab environment where you can work with it. 
<img src="assets/repo_2_docker_example.png" alt="Repo2Docker example" height=3200 width=800 >
