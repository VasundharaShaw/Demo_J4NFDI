## 🔗 Task 2: Mount External Data – at Runtime!

One of the powerful features of **Jupyter4NFDI** is the ability to **mount external storage** into your running JupyterLab session — without restarting.

This works through the **DataMount** extension, already integrated into your session.

---

### 🚀 What Can You Mount?

You can connect to and mount data from:

- **B2DROP**
- **Amazon AWS S3**
- **Nextcloud / OwnCloud**
- **SharePoint**
- Any other **S3**-compatible or **WebDAV**-based storage

> 📁 Once mounted, the data will appear as a normal folder in your JupyterLab file browser — ready to read and write from!

---

### 🧪 Try It Yourself

If you already have access to one of these storage services, go ahead and try mounting your own data!

Use the **mount dialog** from the sidebar in JupyterLab.

📚 You can find full documentation here:  
👉 [DataMount Documentation](https://jsc-jupyter.github.io/jupyterlab-data-mount)


Let’s make data access seamless — directly from where you analyze it!
