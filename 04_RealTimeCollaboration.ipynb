## 👥 Chapter 4: Real-Time Collaboration (RTC)

With **Real-Time Collaboration (RTC)**, you can work together with colleagues **live in the same JupyterLab session**.  
Multiple users can edit notebooks, run code, and explore data simultaneously — just like in a shared Google Doc, but for Jupyter!

---

### 🚀 Step 1: Start Your JupyterLab Session

1. Go to the [Jupyter4NFDI Home](https://hub.nfdi-jupyter.de/hub/home).
2. Launch a JupyterLab session (choose any environment — Repo2Docker, custom image, or preconfigured one).

---

### 🔗 Step 2: Create an RTC Link

1. In the **JupyterLab Configuration**, click the **RTC button**.
2. A dialog will appear with an **invitation link**.
3. Copy the link and share it with your collaborators.

### 👩‍💻 Step 3: Join a Shared Session

Your collaborators can now open the link you shared.  
They will be dropped directly into your running JupyterLab session, with **full access to files, notebooks, and terminals**.

---

### ⚠️ Important Notes

- Anyone with the RTC link has **full access** to your session — including the ability to modify or delete files.  
- RTC links only work while your session is active. Once your JupyterLab stops, all RTC links are revoked.  
- Always **share only with trusted collaborators**.

---

### ✅ What to Try Out

1. Invite a colleague into your RTC session.
2. Open the same notebook together.
3. Try editing different cells — notice how you can both type and execute code in real time.
4. Watch as outputs and results appear **for both of you instantly**.
