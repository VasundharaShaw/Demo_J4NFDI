## 🌍 Chapter 3: FAIR Jupyter – Shareable, Reproducible, Interactive

Let’s explore how **Jupyter4NFDI** enables **FAIR** (Findable, Accessible, Interoperable, Reusable) research and teaching — by letting you create **sharable, interactive Jupyter environments**.

---

### 🛠️ Step-by-Step: Create Your Own FAIR JupyterLab Session

1. Go to the **Jupyter4NFDI Control Panel**: **[https://hub.nfdi-jupyter.de](https://hub.nfdi-jupyter.de)**.
2. Click on **"Add new JupyterLab"**.
3. In the configuration page:
   - Choose **Repo2Docker** in the "Select Version" properties.
   - Enter the following GitHub repository:  
     `binder-examples/requirements`
   - Use `HEAD` as the Git reference.
   - Enable **user persistent storage** by check marking the box "Mount user data" (take `/home/jovyan/work` as your folder).
![Example for configuration](assets/example_create.png)
---

### 💾 Step 2: Preconfigure a DataMount (Optional, but Cool!)

Before launching the JupyterLab session:

1. Click on the **"Storage"** tab in the sidebar.
2. Add a new mount using a template (e.g., B2DROP, AWS, Nextcloud).
3. Set the mount to **readonly** – so others can access but not modify the source.
4. Press start to save your configuration (This will open a new tab with the configured JupyterLab). For later changes: Just press the (new) save button.
5. Go back to the tab with configuration control

---

### 📤 Step 3: Share It!

Once everything is configured, click the **"Share"** button.

This creates a **shareable link** that:

- Launches JupyterLab with your repo and mount setup
- Mounts the external storage **securely in the backend** (no passwords are exposed)
- Reproduces the exact same environment for your colleagues

> ✅ This is the very same feature we used to create the share link used in **this tutorial** ( https://hub.nfdi-jupyter.de/ )!

---

## 🚀 What Can You Do With This?

With this FAIR Jupyter approach, you can:

1. **Reproducible Research**  
   Use a Git repository to analyze data stored externally (e.g., on AWS).  
   Share a link that launches a live, runnable version of your work.

2. **Teaching & Education**  
   Combine **Repo2Docker** with **persistent storage** (`/home/jovyan/work`).  
   Prepare a Git repository with lecture content, exercises, or code examples.  
   Your students can start their environment, work in it, and **come back weeks or months later** — their progress is saved.

3. **Flexible Environments**  
   Don’t want to use Repo2Docker?  
   No problem — use **custom Docker images** or simply pick a **preconfigured environment** for a faster start.
