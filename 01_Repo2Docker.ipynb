## 🧪 Task 1: Try MyBinder – and Understand Its Limits

Let's start by launching a simple JupyterLab session using **MyBinder.org**.

> We'll use a minimal example repository that only contains a `requirements.txt` file to define the Python environment.
> You can find the repository on [github/mybinder-examples/requirements](https://github.com/binder-examples/requirements)

---

### ▶️ Step 1: Launch Binder

Click the link below to launch the example on **MyBinder**:  
🔗 **Launch on MyBinder.org: [https://mybinder.org/v2/gh/binder-examples/requirements/HEAD](https://mybinder.org/v2/gh/binder-examples/requirements/HEAD)**

Wait a few moments while your JupyterLab environment starts.

---

### ✍️ Step 3: Add a File

Before closing the session, create a small Python file to simulate your work:

1. In the JupyterLab file browser, click **"New" → "Text File"**.
2. Rename it to `hello_world.py`.
3. Add the following code:

   ```python
   print("Hello, world!")
   ```

4. Save the file.
5. Run `python hello_world.py` in your terminal

---

### 🧹 Step 4: Close and Reopen

Now, **close the browser tab**.  
Then click the Binder link again to reopen the environment.  
[https://mybinder.org/v2/gh/binder-examples/requirements/HEAD](https://mybinder.org/v2/gh/binder-examples/requirements/HEAD)

> ❌ **Oops – your session is gone!**  
> All your changes and files are lost. Binder environments are **temporary** and reset every time.

---

## ✅ Step 5: Try the Same on Jupyter4NFDI – With Persistence!

Now let’s try the **same repository** using **Jupyter4NFDI** with **persistent storage**.

🔗 **Launch on Jupyter4NFDI: [https://hub.nfdi-jupyter.de/v2/gh/binder-examples/requirements/HEAD](https://hub.nfdi-jupyter.de/v2/gh/binder-examples/requirements/HEAD)**

Repeat the same steps:

1. Create the same `hello_world.py` file.
2. Run `python3 hello_world.py`
3. Close the browser tab.

Then click the link again to reopen the environment.

> 🎉 **Nice! Your notebook and script are still there.**  
> In Jupyter4NFDI, your session includes **persistent storage** — files and changes last for up to **2 hours**, or longer depending on the deployment.

---
