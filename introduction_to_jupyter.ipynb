# Introduction to Jupyter Notebooks

## What are Jupyter Notebooks? 🧮
A Jupyter Notebook is an interactive document that combines code (often Python), text (Markdown, like this format here), mathematics (LaTeX), and visualizations in a single user interface. It is mainly used in data science, research, education, and machine learning.

* Supports many programming languages (mainly Python, but also R, Julia, etc.).

* Consists of so-called “cells” – you can execute code and text sections separately.

* Results (e.g., tables, graphs) appear directly below the code.

📁 File format: .ipynb (JSON-based)

## Further information
In addition to Jupyter Notebooks, there are also Jupyter Labs and Jupyter Hubs. It is not particularly important to know everything in detail, but you may still find it interesting to know what the terms mean. So feel free to read on ;) .

### Jupyter Labs 🧪
JupyterLab is the advanced user interface for Jupyter Notebooks—a kind of “IDE for data work.” Basically, the current area is a JupyterLab.

* Offers a modern, flexible interface with tabs, drag & drop, and panels.

* Supports notebooks, terminals, text editors, file browsers, plot viewers, etc. in one environment.

* Makes working with multiple documents at the same time much more convenient.

🆚 Comparison to Jupyter Notebook:

* Jupyter Notebook is simple and straightforward – JupyterLab is more versatile and modern.

![jupyter_lab_example](assets/jupyter_lab.png)

### JupyterHub 🧑‍🤝‍🧑
JupyterHub is a multi-user system that provides Jupyter Notebooks or JupyterLab centrally for many users—e.g., in schools, universities, research institutions, or companies.

* Offers user authentication (e.g., login with LDAP, GitHub, Google, etc.).
* You may have logged in with your university account.

* Starts a separate notebook environment for each user (often in containers or via SSH).

* Users have direct access to the functions
* No need to install anything yourself
* No administration of the environment required by the user

🛠 Example scenarios:

* An introductory course in web development in which all students use the same environment.

* Research clusters in which users are distributed across powerful computing nodes and can access a wide variety of data sets.
* Share resources
* Share data
* Open up capacities to many researchers
